**Plot a pie chart.**

- Make a pie chart of array x. The fractional area of each wedge is given by x/sum(x).

- The wedges are plotted counterclockwise, by default starting from the x-axis.
  - The plot.pie() function in Matplotlib is used to create pie charts, which visually represent data as proportional slices of a circle. This chart is commonly used for displaying percentages or relative sizes.

```python
animals = 'Frogs', 'Hogs', 'Dogs', 'Logs'
sizes = [15, 30, 45, 10]
plt.pie(sizes, labels=animals)
plt.show()
```

```python
animals = 'Frogs', 'Hogs', 'Dogs', 'Logs'
sizes = [153, 301, 405, 100]
plt.pie(sizes, labels=animals, autopct = '%1.3f%%')
plt.show()
```

```python
animals = 'Frogs', 'Hogs', 'Dogs', 'Logs'
sizes = [153, 301, 405, 100]
plt.pie(sizes, labels=animals, autopct = '%1.3f%%', colors=['red', 'brown', 'aqua', 'saddlebrown'])
plt.show()
```

**Swap label and autopct text positions**

**Hatches
Pass a list of hatch patterns to hatch to set the pattern of each slice

**autopct='%1.1f%%'**: Formats the labels to show the percentage value with one decimal place.

The **autopct** parameter in the matplotlib.pyplot.pie() function is used to format the labels on a pie chart, specifically the percentage values that are displayed on each segment of the pie chart.

**autopct Parameter Explained**
Purpose: The autopct parameter defines how the percentage labels should be formatted on each pie chart segment. This is useful for providing a clear and readable percentage representation of each slice's proportion relative to the whole.

Format String: The value of autopct is a format string that specifies how the percentages should appear. For example, '%1.1f%%' is a format string where:

- **%**: Indicates that you are specifying a format for a value.

- **1.1f**: Indicates a floating-point number with 1 digit before the decimal point and 1 digit after the decimal point.

- **%%**: Adds a literal percent sign (%) to the formatted output.<br>
**autopct (automatic percent): label the wedges with their numeric value. The label will be placed inside the wedge. If autopct is a format string, the label will be fmt % pct**

```python
# Data
sizes = [15, 30, 45, 10]
labels = ['R', 'B', 'G', 'Y']

# Create pie chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['red', 'blue', 'green', 'yellow'])

# Title
plt.title('Simple Pie Chart Example')

# Show plot
plt.show()
```

```python
sizes = [30, 15, 45, 10]
labels = ['Red', 'Blue', 'Green', 'Yellow']
explode = (0.1, 0, 0, 0)  # 'explode' the first slice

plt.pie(sizes, labels=labels, autopct='%1.1f%%', explode=explode, startangle=140, colors=['red', 'blue', 'green', 'yellow'])

plt.title('Exploded Pie Chart Example')

plt.show()
```

```python
sizes = [30, 20, 40, 10]
labels = ['Red', 'Blue', 'Green', 'Yellow']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True, colors=['red', 'blue', 'green', 'yellow'])

plt.title('Pie Chart with Shadow Example')

plt.show()
```

```python
sizes = [30, 40, 20, 10]
labels = ['Red', 'Blue', 'Green', 'Yellow']

# Customize text properties
textprops = {'fontsize': 12, 'fontweight': 'bold', 'color': 'black'}

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['red', 'blue', 'green', 'yellow'], textprops=textprops)

plt.title('Pie Chart with Text Properties')

plt.show()
```

```python
sizes = [30, 40, 20, 10]
labels = ['Red', 'Blue', 'Green', 'Yellow']

#Clockwise vs anticlockwise
textprops = {'fontsize': 12, 'fontweight': 'bold', 'color': 'black'}

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, counterclock = False, colors=['red', 'blue', 'green', 'yellow'], textprops=textprops)

plt.title('Pie Chart with Text Properties')

plt.show()
```

```python
sizes = [30, 40, 20, 10]
labels = ['Red', 'Blue', 'Green', 'Yellow']

#labeldistance
textprops = {'fontsize': 12, 'fontweight': 'bold', 'color': 'black'}

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, labeldistance = 1.3, colors=['red', 'blue', 'green', 'yellow'], textprops=textprops)

plt.title('Pie Chart with Text Properties')

plt.show()
```
