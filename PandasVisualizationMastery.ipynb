```python
df['Math'].plot(kind = 'hist', bins = 5, color = 'purple', edgecolor = 'black', alpha = 0.7, title = "Math scores (5 bins)")
```

## how are bin edges calcualted
1. Determine the range: r = max - min = 95 - 67 = 28
2. determine bin width: bw = r/bins = 28/5= 5.6 [sometimes rounded)
3. calculate bin edges: start at min value and add the bin widht repeatedlly:
    - Bin 1: 67 to 67 + 5.6 [i.e., 67 to 72.6]
    - Bin 2: 72.6 to 78.2........
    - Bin 5: 89.4 to 95

    - Box: Q1 (25th percentile) to Q3 (75th percentile).

    - Whiskers: Typically Q1 - 1.5*IQR to Q3 + 1.5*IQR.
    - Dots: Outliers (scores outside whiskers).

```python
data = [10, 15, 14, 10, 14, 20, 25, 30, 35, 40]
df = pd.Series(data)
df
```

## The “25th percentile” 
    - (also called the first quartile, Q₁) is the value below which 25% of your observations fall. In other words, if you sort your data from smallest to largest, 25% of the data points will be at or below the 25th-percentile value.

    This is the median of the lower half (before the median of the whole dataset). The lower half is: [10, 10, 14, 14, 15]


```python
#calculate the Q1 (25th percentile)
# there are 10 points, 25% of 10 is 2.5, so the 25th percentile lies between the 2nd and 3rd data points [which is 14.0]
Q1 = df.quantile(0.25)
Q1
```

```python
# Q3 (75th percentile: This is the median of the upper half (after the median of the whole dataset). The upper half is: [20, 25, 30, 35, 40]
Q3 = df.quantile(0.75)
Q3
```

```python
#calculate the Inter-Quartile Range (IQR)
IQR = Q3 - Q1
```

```python
# calculate the whiskers [lines below and above the box]
lower_whisker = Q1 - 1.5 * IQR
upper_whisker = Q3 + 1.5 * IQR
```

```python
# identify outliers
outliers = df[(df < lower_whisker) | (df > upper_whisker)]
```

## Box plot therefore show:
    - The box: The middle 50% of the data, from Q1 (25th percentile) to Q3 (75th percentile).
    - The line inside the box: The median (Q2).
    - The whiskers: Extend from Q1 - 1.5 * IQR to Q3 + 1.5 * IQR, showing the range of data excluding outliers.
    - Outliers: Any data points outside the whiskers are plotted as dots.

```python
df.plot(kind = 'box')
```

## What will be displayed:
    - Box: The box will represent the range from Q1 (25th percentile) to Q3 (75th percentile).
    - Line inside the box: The median value (Q2), which is 17.5 in the example.
    - Whiskers: The lines extending from the box to the minimum and maximum values within the whisker boundaries (i.e., 10 and 40 in the example).
    - Outliers: Any points beyond the whiskers (none in this example).

```python
# Updated dataset with an outlier (100)
data_with_outlier = [10, 15, 14, 10, 14, 20, 25, 30, 35, 40, 100]
df = pd.Series(data_with_outlier)

# Plot the box plot
df.plot(kind='box')

# Add title and labels
plt.title('Box Plot of Data with Outlier')
plt.xlabel('Scores')

# Show the plot
plt.show()
```

## Explanation:
**Outlier**: The value 100 is an outlier, and it will be shown as a separate dot beyond the whiskers.

## Box Plot Details:
    - The box will now represent the interquartile range between Q1 (25th percentile) and Q3 (75th percentile).
    - The whiskers will still represent the range of data within 1.5 * IQR from Q1 and Q3.
    - The outlier will be shown as a point outside the whisker range.

## Expected Result:
    - The whiskers will still extend from the minimum (10) to the maximum value within the whisker range (40).
    - The box will cover the middle 50% of the data, from Q1 (14) to Q3 (30).
    - The outlier at 100 will be plotted as a separate dot outside the whiskers.
